```python
# Data Loading

import pandas as pd
from pathlib import Path

# Find CSV files in the project folder
csv_files = Path("..").rglob("*.csv")

# Store datasets
data = {}

# Load each CSV file
for file in csv_files:
    dataset_name = file.stem
    data[dataset_name] = pd.read_csv(file)

# Display loaded datasets
for name, df in data.items():
    print(f"{name}: {df.shape}")
```

```python
# Dataset Profiling

for name, df in data.items():
    print(f"\n{'='*50}")
    print(name.upper())
    print(f"\n{'='*50}")

    print("Rows:", df.shape[0])
    print("Column:",df.shape[1])
    print("Missing Values:", df.isna().sum().sum())
    print("Duplicate rows:", df.duplicated().sum())
```

```python
# Numeric Profiling

for name, df in data.items():
    print(f"\n{'='*50}")
    print(name.upper())
    print(f"\n{'='*50}")

    display(df.describe().T)
```

```python
# Identify Data Quality Issue

for name, df in data.items():
    print(f"\n{'='*50}")
    print(name.upper())
    print(f"\n{'='*50}")

    display(df.describe().T)

    missing=df.isna().sum()
    missing_pct = (missing/len(df)*100).round(2)

    profile=pd.DataFrame({
        "missing_count": missing,
        "missing_pct": missing_pct
    })

    display(profile[profile["missing_count"]>0])
```

```python
# Data quality issue summary
for name, df in data.items():
    print(f"\n{'='*50}")
    print(name.upper())
    print(f"\n{'='*50}")

    print("Duplicate rows:", df.duplicated().sum())
    print("Total Missing Values:", df.isna().sum().sum())
```
