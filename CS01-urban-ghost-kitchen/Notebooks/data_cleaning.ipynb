```python
# Numeric Cleaning

# 1. Fix negative order_value
negative_order_value = clean_data["orders"]["order_value"] < 0

if negative_order_value.sum() > 0:
    clean_data["orders"].loc[negative_order_value, "order_value"] = (
        clean_data["orders"]
        .loc[negative_order_value, "order_value"]
        .abs()
    )
    print("Negative order_value values corrected")
else:
    print("No negative order_value values found")
```

```python
# 2. Fix negative delivery_fee
negative_delivery_fee = clean_data["orders"]["delivery_fee"] < 0

if negative_delivery_fee.sum() > 0:
    clean_data["orders"].loc[negative_delivery_fee, "delivery_fee"] = (
        clean_data["orders"]
        .loc[negative_delivery_fee, "delivery_fee"]
        .abs()
    )
    print("Negative delivery_fee values corrected")
else:
    print("No negative delivery_fee values found")
```

```python
# 3. Fix negative unit_price
negative_unit_price = clean_data["menu_items"]["unit_price"] < 0

if negative_unit_price.sum() > 0:
    clean_data["menu_items"].loc[negative_unit_price, "unit_price"] = (
        clean_data["menu_items"]
        .loc[negative_unit_price, "unit_price"]
        .abs()
    )
    print("Negative unit_price values corrected")
else:
    print("No negative unit_price values found")
```

```python
# 6. Handle zero delivery time
zero_delivery_time = clean_data["orders"]["delivery_time_min"] == 0

if zero_delivery_time.sum() > 0:
    clean_data["orders"].loc[zero_delivery_time, "delivery_time_min"] = (
        clean_data["orders"]["delivery_time_min"]
        .median()
    )
    print("Zero delivery time values corrected")
else:
    print("No zero delivery time values found")
```

```python
# Review extreme outliers

numeric_columns = {
    "orders": [
        "order_value",
        "discount",
        "delivery_fee",
        "delivery_distance_km",
        "delivery_time_min"
    ],
    "menu_items": [
        "quantity",
        "unit_price"
    ],
    "kitchens": [
        "monthly_rent",
        "staff_count"
    ]
}

for name, columns in numeric_columns.items():
    print(f"\n{'=' * 50}")
    print(name.upper())
    print(f"{'=' * 50}")

    for column in columns:
        q1 = clean_data[name][column].quantile(0.25)
        q3 = clean_data[name][column].quantile(0.75)
        iqr = q3 - q1

        lower_bound = q1 - 1.5 * iqr
        upper_bound = q3 + 1.5 * iqr

        outlier_count = (
            (clean_data[name][column] < lower_bound) |
            (clean_data[name][column] > upper_bound)
        ).sum()

        print(
            f"{column}: {outlier_count} outliers "
            f"(lower: {lower_bound:.2f}, upper: {upper_bound:.2f})"
        )
```
