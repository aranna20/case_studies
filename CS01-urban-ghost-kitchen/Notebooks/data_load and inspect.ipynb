```python
# Data Loading

import pandas as pd
from pathlib import Path

# Find CSV files in the project folder
csv_files = Path("..").rglob("*.csv")

# Store datasets
data = {}

# Load each CSV file
for file in csv_files:
    dataset_name = file.stem
    data[dataset_name] = pd.read_csv(file)

# Display loaded datasets
for name, df in data.items():
     print(f"{name}: {df.shape}")
```

```python
# Data Inspect

for name, df in data.items():
    print(f"\n{'='*50}")
    print(name.upper())
    print(f"\n{'='*50}")

    print("Rows:", df.shape[0])
    print("Column:",df.shape[1])
```

```python
# Preview the data

for name, df in data.items():
    print(f"\n{'='*50}")
    print(name.upper())
    print(f"\n{'='*50}")

    display(df.head())
```

```python
# Summarise datatypes

for name, df in data.items():
    print(f"\n{'='*50}")
    print(name.upper())
    print(f"\n{'='*50}")

    print(df.dtypes)
```

```python
# Check Missing Values

for name, df in data.items():
    print(f"\n{'='*50}")
    print(name.upper())
    print(f"\n{'='*50}")

    print(df.isna().sum())
```

```python
# Duplicacy checking for Primary Key

pk= {
    "customers":"customer_id",
    "kitchens":"kitchen_id",
    "menu_items":"item_id",
    "orders":"order_id"
}

for name, column in pk.items():
    duplicate_counts=data[name][column].duplicated().sum()

    print(f"{name}.{column}:{duplicate_counts} duplicate values")
```

```python
# Check Unique Values for Categorical Columns

cat_cols={
    "customers":["city","customer_segment"],
    "kitchens":["zone"],
    "menu_items":["category"],
    "orders":["order_status"]
}

for name, columns in cat_cols.items():
    print(f"\n{'='*50}")
    print(name.upper())
    print(f"{'='*50}")

    for column in columns:
        print(f"\n{column}:")
        print(data[name][column].value_counts(dropna=False))    
```

```python
# Numeric Data Inspections

for name, df in data.items():
    print(f"\n{'='*50}")
    print(name.upper())
    print(f"\n{'='*50}")

    display(df.describe().T)
```
